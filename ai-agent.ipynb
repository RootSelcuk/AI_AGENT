```python

from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model_name="llama-3.3-70b-versatile",

)
```

```python
response= llm.invoke("model özellikleri nedir?")
response.content
```

```python
from langchain.chat_models import init_chat_model

from langchain.agents import create_agent

modelgroq = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")


agent=create_agent(model=modelgroq)
```

```python
from langchain.messages import HumanMessage
response = agent.invoke({"messages":[HumanMessage(content="Paris'in başkenti neresi")]})
print(response['messages'][-1].content)
```
